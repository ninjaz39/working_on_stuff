```python
def save_ndarray(array: np.ndarray, name: str) -> None:
    """
    Saves an n-dimensional numpy array to a file.

    File format:
      Line 1: number of dimensions (n)
      Line 2: shape of the array (space-separated)
      Remaining lines: flattened array data (space-separated)

    Args:
        array: The n-dimensional numpy array to save.
        name:  The output filename (e.g. "my_array.txt").
    """
    n = array.ndim
    shape = array.shape
    flat = array.flatten()

    with open(name, "w") as f:
        # Line 1: number of dimensions
        f.write(f"{n}\n")
        # Line 2: shape
        f.write(" ".join(str(s) for s in shape) + "\n")
        # Remaining: flattened data
        f.write(" ".join(str(v) for v in flat) + "\n")
```

```python
def plot_solved_domain(point, domain, val):
    curr_dim = domain
    for i in range(len(point)- 1):
        curr_dim = curr_dim[point[i]]
    curr_dim[point[-1]] += val

import numpy as np

def support_vec(shape1, shape2, d):
    """
    Support point of Minkowski difference A - B
    """
    return (shape1.support(d) - shape2.support(-d))

def del_y(simplex):
    ans = np.zeros(len(simplex))
    
    for index in range(len(simplex)):
        ans[index] = del_yj(simplex, index)
    return ans

def del_yj(simplex, index_j):
    if len(simplex) == 1:
        return 1
    
    ans = 0
    vj = simplex[index_j]
    simplex_i =  simplex.copy()
    simplex_i.pop(index_j)
    k = simplex_i[0]
    for i, j in enumerate(simplex_i):
        del_yi = del_yj(simplex_i, i)
        ans += del_yi * (j@(k-vj))
    return ans

def nearest_simplex(simplex):
    if len(simplex) == 1:
        return simplex, None
    
    inside = True
    del_simplex = del_y(simplex)
    for i in del_simplex:
        if i<=0:
            inside = False
            break
    if inside:
        return simplex, del_simplex

    for i in range(len(simplex)-1):
        new_simplex = simplex.copy()
        new_simplex.pop(i)
        new_simplex, del_simplex = nearest_simplex(new_simplex)
        if type(del_simplex) != type(None):
            return new_simplex, del_simplex
    
    return [simplex[-1]], None

def distance_subalgorithm(simplex):
    new_simplex, del_simplex = nearest_simplex(simplex)
    if type(del_simplex) == type(None):
        return new_simplex, -1 * new_simplex[0], np.linalg.norm(new_simplex[0])
    
    v = np.zeros(new_simplex[0].shape)
    for i in range(len(new_simplex)):
        v -= new_simplex[i]*del_simplex[i]
    
    del_simplex_val = sum(del_simplex)
    v /= del_simplex_val
    return new_simplex, v, np.linalg.norm(v)

def gjk_distance(shape1, shape2, simplex, distPrev = None, alg = True):#alg = True means it uses DS, False uses BP
    """Calculate the distance between two convex shapes using the GJK algorithm."""
    new_simplex, v, dist = distance_subalgorithm(simplex)
    support = support_vec(shape1, shape2, v)
    #print(simplex, "simplex", support, 'support', v, 'v', distPrev, dist, 'dist', gk(-v, support))

    if gk(v, support) == 0:
        return dist#, v, new_simplex, True

    if (not distPrev) or dist < distPrev:
        new_simplex.append(support)
        return gjk_distance(shape1, shape2, new_simplex, dist)
    
    if not alg: #backup procedure used and tolerance not satisfied
        return dist#, v, new_simplex, alg #backup procedure used and tolerance not satisfied
    
    return gjk_distance(shape1, shape2, simplex, distPrev, False)
    
def gk(v, s):
    return np.linalg.norm(v)**2 + s@v

def dist_to_boundary(x, obstacles, boundary = None):
    if boundary:
        dist = boundary.radius - np.linalg.norm(x-boundary.center)
        if dist<0:
            return 0, -1
    else:
        dist = float('inf')
    obstacle_hit = -1
    start_d = np.random.uniform(0,1,len(x))
    x = Sphere(x, 0)
    for index, obstacle in enumerate(obstacles):
        dist_o = gjk_distance(x, obstacle, [support_vec(x, obstacle, start_d)])
        if dist > dist_o:
            dist = dist_o
            obstacle_hit = index
    return dist, obstacle_hit



from tqdm import tqdm
import numpy as np

def solve_domain_dist_to_b(
    domain,
    domain_shape,
    obstacles,
    min_corner,
    dim,
    curr,
    pbar
):

    if len(curr) == dim:
        curr = np.array(curr)

        val, obstacle_hit = dist_to_boundary(curr + min_corner, obstacles)
        plot_solved_domain(curr, domain, val)

        pbar.update(1)
        return

    for i in range(domain_shape[0]):
        nxt = curr.copy()
        nxt.append(i)

        solve_domain_dist_to_b(
            domain,
            domain_shape[1:],
            obstacles,
            min_corner,
            dim,
            nxt,
            pbar
        )
```

```python
from IPython.display import HTML
def visualize_moving_obstacle(
    corners_4d,
    steps=50,
    interval=100,
):
    """
    corners_4d: (16, 4) array-like, the 16 corners of a 4D box.
                Each corner is [x, y, z, t].
                Corners 0-7  are the t0 face.
                Corners 8-15 are the t1 face.

    Example
    -------
    corners = [
        [100, 50, 50, 10], [100, 50, 52, 10], [100, 54, 50, 10], [100, 54, 52, 10],
        [105, 50, 50, 10], [105, 50, 52, 10], [105, 54, 50, 10], [105, 54, 52, 10],
        [190, 50, 50, 100],[190, 50, 52, 100],[190, 54, 50, 100],[190, 54, 52, 100],
        [195, 50, 50, 100],[195, 50, 52, 100],[195, 54, 50, 100],[195, 54, 52, 100],
    ]
    visualize_moving_obstacle(corners)
    """
    corners_4d = np.array(corners_4d, dtype=float)  # (16, 4)
    assert corners_4d.shape == (16, 4), f"Expected (16, 4), got {corners_4d.shape}"

    # split into t0 and t1 faces
    face_t0 = corners_4d[:8]   # (8, 4)
    face_t1 = corners_4d[8:]   # (8, 4)

    t0 = face_t0[0, 3]
    t1 = face_t1[0, 3]
    ts = np.linspace(0, 1, steps)  # interpolation parameter

    # interpolate corners across time: (steps, 8, 3)
    xyz_t0 = face_t0[:, :3]  # (8, 3)
    xyz_t1 = face_t1[:, :3]  # (8, 3)
    all_corners = xyz_t0[None] + ts[:, None, None] * (xyz_t1 - xyz_t0)[None]  # (steps, 8, 3)

    edges = [
        (0,1),(0,2),(0,4),
        (1,3),(1,5),
        (2,3),(2,6),
        (3,7),
        (4,5),(4,6),
        (5,7),
        (6,7),
    ]

    # global bounds for stable axes
    all_xyz = np.vstack([xyz_t0, xyz_t1])
    pad = 1.0
    x_min, y_min, z_min = all_xyz.min(axis=0) - pad
    x_max, y_max, z_max = all_xyz.max(axis=0) + pad

    fig = plt.figure(figsize=(8, 8))
    ax  = fig.add_subplot(111, projection='3d')
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(y_min, y_max)
    ax.set_zlim(z_min, z_max)
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("Z")

    lines = [ax.plot([], [], [], lw=2, color='steelblue')[0] for _ in edges]

    progress_ax = fig.add_axes([0.15, 0.05, 0.7, 0.03])
    progress_ax.set_xlim(t0, t1)
    progress_ax.set_ylim(0, 1)
    progress_ax.set_xlabel("t")
    progress_ax.set_yticks([])
    progress_bar, = progress_ax.plot([], [], lw=6, color='steelblue')
    progress_ax.axvline(t0, color='gray', lw=1)
    progress_ax.axvline(t1, color='gray', lw=1)

    def update(frame):
        corners = all_corners[frame]  # (8, 3)
        t = t0 + ts[frame] * (t1 - t0)

        for line, (i, j) in zip(lines, edges):
            p1, p2 = corners[i], corners[j]
            line.set_data([p1[0], p2[0]], [p1[1], p2[1]])
            line.set_3d_properties([p1[2], p2[2]])

        alpha = ts[frame]
        color = (alpha, 0.3, 1 - alpha)
        for line in lines:
            line.set_color(color)

        progress_bar.set_data([t0, t], [0.5, 0.5])
        ax.set_title(f"3D Obstacle   t = {t:.2f}  /  {t1:.2f}")

        return lines + [progress_bar]

    anim = FuncAnimation(fig, update, frames=steps, interval=interval, blit=False)
    fig.subplots_adjust(bottom=0.15)
    plt.close()
    return HTML(anim.to_jshtml())
```

```python
from IPython.display import HTML
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

def visualize_moving_obstacle(
    corners_4d,
    path=None,
    steps=50,
    interval=100,
):
    """
    corners_4d: (16, 4) array-like, the 16 corners of a 4D box.
    path:       (N, 4) array-like of [x, y, z, t] points to plot as a trail.
                Only points within [t0, t1] are shown, revealed progressively.

    Example
    -------
    visualize_moving_obstacle(corners, path=path_array)
    """
    corners_4d = np.array(corners_4d, dtype=float)
    assert corners_4d.shape == (16, 4), f"Expected (16, 4), got {corners_4d.shape}"

    face_t0 = corners_4d[:8]
    face_t1 = corners_4d[8:]

    t0 = face_t0[0, 3]
    t1 = face_t1[0, 3]
    ts = np.linspace(0, 1, steps)

    xyz_t0 = face_t0[:, :3]
    xyz_t1 = face_t1[:, :3]
    all_corners = xyz_t0[None] + ts[:, None, None] * (xyz_t1 - xyz_t0)[None]  # (steps, 8, 3)

    # parse path
    path_xyz = None
    path_t   = None
    if path is not None:
        path = np.array(path, dtype=float)  # (N, 4)
        path_xyz = path[:, :3]
        path_t   = path[:, 3]

    edges = [
        (0,1),(0,2),(0,4),
        (1,3),(1,5),
        (2,3),(2,6),
        (3,7),
        (4,5),(4,6),
        (5,7),
        (6,7),
    ]

    # global bounds including path
    all_xyz = np.vstack([xyz_t0, xyz_t1])
    if path_xyz is not None:
        all_xyz = np.vstack([all_xyz, path_xyz])
    pad = 1.0
    x_min, y_min, z_min = all_xyz.min(axis=0) - pad
    x_max, y_max, z_max = all_xyz.max(axis=0) + pad

    fig = plt.figure(figsize=(8, 8))
    ax  = fig.add_subplot(111, projection='3d')
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(y_min, y_max)
    ax.set_zlim(z_min, z_max)
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("Z")

    lines = [ax.plot([], [], [], lw=2, color='steelblue')[0] for _ in edges]

    # path trail: revealed up to current t
    trail_line,     = ax.plot([], [], [], lw=1.5, color='limegreen', alpha=0.8, label='path')
    trail_dot,      = ax.plot([], [], [], 'o', color='limegreen', ms=5)  # current position
    if path is not None:
        ax.legend(loc='upper left')

    progress_ax = fig.add_axes([0.15, 0.05, 0.7, 0.03])
    progress_ax.set_xlim(t0, t1)
    progress_ax.set_ylim(0, 1)
    progress_ax.set_xlabel("t")
    progress_ax.set_yticks([])
    progress_bar, = progress_ax.plot([], [], lw=6, color='steelblue')
    progress_ax.axvline(t0, color='gray', lw=1)
    progress_ax.axvline(t1, color='gray', lw=1)

    def update(frame):
        corners = all_corners[frame]  # (8, 3)
        t = t0 + ts[frame] * (t1 - t0)

        # update box edges
        for line, (i, j) in zip(lines, edges):
            p1, p2 = corners[i], corners[j]
            line.set_data([p1[0], p2[0]], [p1[1], p2[1]])
            line.set_3d_properties([p1[2], p2[2]])

        alpha = ts[frame]
        color = (alpha, 0.3, 1 - alpha)
        for line in lines:
            line.set_color(color)

        # reveal path up to current t
        if path_xyz is not None:
            mask = path_t <= t
            if mask.any():
                visible = path_xyz[mask]
                trail_line.set_data(visible[:, 0], visible[:, 1])
                trail_line.set_3d_properties(visible[:, 2])
                # current position dot
                trail_dot.set_data([visible[-1, 0]], [visible[-1, 1]])
                trail_dot.set_3d_properties([visible[-1, 2]])

        progress_bar.set_data([t0, t], [0.5, 0.5])
        ax.set_title(f"3D Obstacle   t = {t:.2f}  /  {t1:.2f}")

        return lines + [trail_line, trail_dot, progress_bar]

    anim = FuncAnimation(fig, update, frames=steps, interval=interval, blit=False)
    fig.subplots_adjust(bottom=0.15)
    plt.close()
    return HTML(anim.to_jshtml())
```
